```python
# プログラミング言語:python

'''
　１.以下のイベントを作成する
　２.各属性にアクセスし、値を出力する
'''

# クラス名:Event
class Event:

# 属性: name , date , locaton
  def __init__(self,name,date,location):

# コンストラクタ
    self.name = name
    self.date = date
    self.location =location

event = Event("プログラミング初心者向けセミナー","2026/11/01","東京高裁フォーラム")

print ( event.name)
print (event.date)
print (event.location)
```

```python
# プログラミング言語:python

'''
　１.以下の人データベースを作成する
　２.各属性にアクセスし、値を出力する
'''

# クラス名
class Human:

# 属性 name , age
# コンストラクタ
  def __init__(self, name, age):
    self.name = name
    self.age  = age

# インスタンスの作成
printinfo = Human("サムライ太郎", 20)

# 各属性にアクセスし、値を出力する
print(f"名前：{printinfo.name}")
print(f"年齢：{str( printinfo.age ) }歳")
```

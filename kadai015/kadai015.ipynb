```python
# プログラミング言語:python

'''
　１.以下の人データベースを作成する
　２.各属性にアクセスし、値を出力する
'''

# クラス名
class Human:

# 属性 name , age
  def __init__(printinfo, name, age):
# コンストラクタ
    printinfo.name = name
    printinfo.age  = age

# インスタンスの作成
human = Human("サムライ太郎", 20)

# 各属性にアクセスし、値を出力する
print(f"名前：{human.name}")
print(f"年齢：{str( human.age ) }歳")
```

```python
class Human:
  def __init__(self,name,age):
    self.name = name
    self.age = age

  def check_adult(self):
    if self.age >= 20:
      print(f"{self.name}は大人です")
    else:
      print(f"{self.name}は子供です")

human = Human("サムライ太郎",20)
human.check_adult()

human = Human("サムライ二郎",17)
human.check_adult()

```

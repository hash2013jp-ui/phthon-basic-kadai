```python
class Human:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def check_adult(self):
    if self.age >= 20:
      print(f"{self.name}は大人です")
    else:
      print(f"{self.name}は子供です")

profiles = [
  {"name": "サムライ太郎", "age": 22},
  {"name": "サムライ二郎", "age": 20},
  {"name": "サムライ三郎", "age": 19}
]

for profile in profiles:
  human = Human(profile["name"], profile["age"])
  human.check_adult()
```
